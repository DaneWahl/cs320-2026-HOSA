## Workout Progress Logger App

This application allows you to log your daily workout progress and view past entries.

```python
# Define the filename for the workout log
WORKOUT_LOG_FILE = 'workout_log.txt'

def write_workout_log(entry):
  """Writes a workout entry to the log file."""
  with open(WORKOUT_LOG_FILE, 'a') as file_object:
    file_object.write(entry + '\n')
  print("Workout logged successfully!")
```

```python
def read_workout_log():
  """Reads and prints all entries from the workout log file."""
  try:
    with open(WORKOUT_LOG_FILE, 'r') as file_object:
      contents = file_object.read()
      if contents:
        print("\n--- Your Workout Log ---")
        print(contents.strip())
        print("------------------------")
      else:
        print("Your workout log is currently empty.")
  except FileNotFoundError:
    print("No workout log found yet. Start by logging your first workout!")
```

```python
import datetime

def run_workout_app():
  """Runs the interactive workout logger application."""
  while True:
    print("\nWhat would you like to do?")
    print("1. Log a new workout")
    print("2. View all workout entries")
    print("3. Exit")

    choice = input("Enter your choice (1, 2, or 3): ")

    if choice == '1':
      workout_details = input("Enter your workout details (e.g., '2024-07-20: Ran 5km, lifted weights'): ")
      # Add a timestamp to the entry if not already present in the user input
      if not workout_details.startswith(str(datetime.date.today())):
          timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
          entry = f"{timestamp}: {workout_details}"
      else:
          entry = workout_details
      write_workout_log(entry)
    elif choice == '2':
      read_workout_log()
    elif choice == '3':
      print("Exiting workout logger. Stay fit!")
      break
    else:
      print("Invalid choice. Please enter 1, 2, or 3.")

# Run the application
run_workout_app()
```

```python
def numbers(limit):
  i = 0
  numbers = []

  while i < limit:
    numbers.append(i)
    i = i + 1
  return numbers

user_limit = int(input("Give a limit: "))
print(numbers(user_limit))
```
