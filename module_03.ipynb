## Library Card User Management Program

This program will allow you to manage library card holder information, including adding new users, editing existing user details, and displaying all registered users. User names will serve as unique identifiers, preventing duplicate entries for the same name.

```python
# Initialize the dictionary to store library user information
# The key will be the user's name (string), and the value will be a dictionary
# containing 'address' and 'phone_number'.
library_users = {}
```

### Add New User Function

This function adds a new user to the `library_users` dictionary. It checks if a user with the same name already exists to prevent duplicates. If the name exists, it will not add the user.

```python
def add_user(name, address, phone_number):
    """Adds a new user to the library_users dictionary."""
    if name in library_users:
        print(f"User '{name}' already exists. Cannot add duplicate user.")
    else:
        library_users[name] = {
            'address': address,
            'phone_number': phone_number
        }
        print(f"User '{name}' added successfully.")
```

### Edit User Information Function

This function allows you to update the address and/or phone number for an existing user. It requires the user's name to identify the record.

```python
def edit_user(name, new_address=None, new_phone_number=None):
    """Edits the address or phone number of an existing user."""
    if name in library_users:
        if new_address:
            library_users[name]['address'] = new_address
            print(f"Address for '{name}' updated to '{new_address}'.")
        if new_phone_number:
            library_users[name]['phone_number'] = new_phone_number
            print(f"Phone number for '{name}' updated to '{new_phone_number}'.")
        if not new_address and not new_phone_number:
            print(f"No updates provided for user '{name}'.")
    else:
        print(f"User '{name}' not found.")
```

### Display All Users Function

This function prints the details of all users currently registered in the `library_users` dictionary.

```python
def display_users():
    """Displays all registered library users and their details."""
    if not library_users:
        print("No library users registered yet.")
        return

    print("\n--- Current Library Users ---")
    for name, details in library_users.items():
        print(f"Name: {name}\n  Address: {details['address']}\n  Phone: {details['phone_number']}\n")
    print("-----------------------------")
```

### Program Demonstration

Here's a demonstration of how to use the functions we've defined:

```python
# Add some users
add_user("Alice Smith", "123 Main St", "555-1111")
add_user("Bob Johnson", "456 Oak Ave", "555-2222")
add_user("Charlie Brown", "789 Pine Ln", "555-3333")

# Try adding a duplicate user (by name)
add_user("Alice Smith", "New Address", "555-0000")

# Display all users
display_users()

# Edit an existing user's information
edit_user("Bob Johnson", new_address="999 Elm Rd")
edit_user("Charlie Brown", new_phone_number="555-4444")
edit_user("Alice Smith", new_address="321 High St", new_phone_number="555-5555")

# Try editing a non-existent user
edit_user("David Lee", new_address="Non Existent")

# Display users after edits
display_users()
```
